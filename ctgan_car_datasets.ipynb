```python
from katabatic.models.ctgan.models import CTGANModel
from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from utils import discretize_preprocess
import os
```

```python
# Create necessary directories
os.makedirs('discretized_data', exist_ok=True)
os.makedirs('sample_data/car', exist_ok=True)
os.makedirs('synthetic/car/ctgan', exist_ok=True)
os.makedirs('Results/car', exist_ok=True)

print("Directory structure created successfully!")
```

```python
print("Step 1: Preprocessing car dataset...")
# Preprocess the car dataset
discretize_preprocess(
    file_path="raw_data/car.csv",
    output_path="discretized_data/car.csv",
    bins=10,
    strategy='uniform'
)
print("Preprocessing complete!")
```

```python
print("\nStep 2: Running CTGAN pipeline...")
# Create pipeline with CTGAN model
pipeline = TrainTestSplitPipeline(model=CTGANModel)

# Run the complete workflow
results = pipeline.run(
    input_csv='discretized_data/car.csv',
    output_dir='sample_data/car',
    synthetic_dir='synthetic/car/ctgan',
    real_test_dir='sample_data/car'
)

print("\nPipeline complete!")
print("\nCTGAN TSTR Results:")
print(results)

print("\nResults saved to: Results/car/ctgan_tstr.csv")
```
