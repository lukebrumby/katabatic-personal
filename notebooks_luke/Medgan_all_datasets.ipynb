```python
from katabatic.models.medgan.models import MEDGAN
from katabatic.pipeline.train_test_split.pipeline import TrainTestSplitPipeline
from utils import discretize_preprocess
import os
import pandas as pd
from datetime import datetime
```

```python
# Define all datasets to process
datasets = ['adult', 'car', 'magic', 'nursery', 'shuttle']

# Preprocessing parameters
bins = 10
strategy = 'uniform'

# Storage for all results
all_results = {}
```

```python
# Create necessary directories for all datasets
os.makedirs('discretized_data', exist_ok=True)
os.makedirs('Results', exist_ok=True)

for dataset in datasets:
    os.makedirs(f'sample_data/{dataset}', exist_ok=True)
    os.makedirs(f'synthetic/{dataset}/medgan', exist_ok=True)
    os.makedirs(f'Results/{dataset}', exist_ok=True)
    
print("Directory structure created successfully!")
```

```python
# Process each dataset
for dataset in datasets:
    print("\n" + "="*80)
    print(f"Processing {dataset.upper()} Dataset")
    print("="*80)
    
    try:
        # Step 1: Preprocessing
        print(f"\nStep 1: Preprocessing {dataset} dataset...")
        discretize_preprocess(
            file_path=f"raw_data/{dataset}.csv",
            output_path=f"discretized_data/{dataset}.csv",
            bins=bins,
            strategy=strategy
        )
        print(f"Preprocessing complete for {dataset}!")
        
        # Step 2: Run MedGAN Pipeline
        print(f"\nStep 2: Running MedGAN pipeline for {dataset}...")
        pipeline = TrainTestSplitPipeline(model=MEDGAN)
        
        results = pipeline.run(
            input_csv=f'discretized_data/{dataset}.csv',
            output_dir=f'sample_data/{dataset}',
            synthetic_dir=f'synthetic/{dataset}/medgan',
            real_test_dir=f'sample_data/{dataset}'
        )
        
        # Store results
        all_results[dataset] = results
        
        print(f"\n✓ Pipeline complete for {dataset}!")
        print(f"\nMedGAN TSTR Results for {dataset}:")
        print(results)
        print(f"\nResults saved to: Results/{dataset}/medgan_tstr.csv")
        
    except Exception as e:
        print(f"\n✗ Error processing {dataset}: {str(e)}")
        all_results[dataset] = None

print("\n" + "="*80)
print("All datasets processed!")
print("="*80)
```
